```python
import numpy as np

# Pauli matrices and identity
sx = np.array([[0,1],[1,0]])
sz = np.array([[1,0],[0,-1]])
I = np.eye(2)

# Build Hamiltonian for n spins
def transverse_field_Ising(n):
    H = np.zeros((2**n, 2**n))
    for k in range(n-1):
        op = 1
        for i in range(n):
            if i == k:      op = np.kron(op, sz)
            elif i == k+1:  op = np.kron(op, sz)
            else:            op = np.kron(op, I)
        H += op
    for k in range(n):
        op = 1
        for i in range(n):
            op = np.kron(op, sx if i == k else I)
        H -= op
    return H

H10 = transverse_field_Ising(10)
E, _ = np.linalg.eigh(H10)
print(E[0])
```
