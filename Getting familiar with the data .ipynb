# Getting familiar with the coco library in python 

The goal of this cell is to get knowledge on how the data looks like. Here i am picking a specific data: the dim 2 and the first function.

I also want to see how it presents the data, here we have: 
- Targets (precision reached)
- Fevals (nbr of evaluations before reaching a acertain preciison) 
- Instances (this is the numer of times we repeat the experiment for a given function and a given algo and a goven dimension, we are simply imputing differnt starting values for the function)  

Tragets: the grid of target precisions (Δf = f(x) − f*), from 1e2 down to 1e-8.
One value per row of the table.

```python
import cocopp
dsl = cocopp.load("bbob/2009/*")
```

```python
dd = dsl.dictByDimFunc()

ds_list = dd[2][1]        # list of DataSet for dim=2, f1
ds0 = ds_list[0]          # pick ONE dataset (one algo)

# Option A — use the precomputed table (targets × runs)
targets  = ds0.evals[:, 0]
fevals   = ds0.evals[:, 1:]      # (#targets, #instances)
instances = ds0.instancenumbers  # usually length 15
targets
```

Fevals: (shape = n_targets × 15) = for each target (row) and each instance (column), the number of function evaluations needed to first hit that target.
1s at the top rows are normal: coarse targets like 1e2 are often already met at the first evaluation.
Unreached targets would be np.inf

```python
fevals
```
